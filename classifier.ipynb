```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
v=TfidfVectorizer()
```

```python
X_subject = v.fit_transform(df["subject"])
```

```python
X_body=v.fit_transform(df['body'])
```

```python
X_sender_email=v.fit_transform(df['sender_email'])
```

```python
X_sender_domain=v.fit_transform(df['sender_domain'])
```

```python
X_claimed_company=v.fit_transform(df['claimed_company'])
```

```python
X_company_domain=v.fit_transform(df['company_domain'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
df["subject_embedding"] = list(X_subject.toarray())
df["body_embedding"]=list(X_body.toarray())
df["sender_email_embedding"]=list(X_sender_email.toarray())
df["sender_domain_embedding"]=list(X_sender_domain.toarray())
df["claimed_company_embedding"]=list(X_claimed_company.toarray())
df["company_domain_emb"]=list(X_company_domain.toarray())
```

```python
df.head()
```

```python
import numpy as np

X = np.hstack([
    np.vstack(df["subject_embedding"].to_numpy()),
    np.vstack(df["body_embedding"].to_numpy()),
    np.vstack(df["sender_email_embedding"].to_numpy()),
    np.vstack(df["sender_domain_embedding"].to_numpy()),
    np.vstack(df["claimed_company_embedding"].to_numpy()),
    np.vstack(df["company_domain_emb"].to_numpy()),
])
```

```python
X_flags = df[
    [
        "domain_match",
        "free_or_third_party_domain",
        "has_link",
        "has_attachment",
        "mentions_money",
        "mentions_urgency",
        "mentions_personal_info",
        "job_offer_without_interview",
    ]
].astype("float32").to_numpy()
```

```python
import numpy as np

X = np.hstack([
    np.vstack(df["subject_embedding"].to_numpy()),
    np.vstack(df["body_embedding"].to_numpy()),
    np.vstack(df["sender_email_embedding"].to_numpy()),
    np.vstack(df["sender_domain_embedding"].to_numpy()),
    np.vstack(df["claimed_company_embedding"].to_numpy()),
    np.vstack(df["company_domain_emb"].to_numpy()),
    X_flags
])
```

```python
Y=df.label
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=10)
```

```python
len(X_train)
```

```python
model=LogisticRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
Y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(Y_test,Y_pred))
```
